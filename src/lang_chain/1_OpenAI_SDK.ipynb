Practicing OPENAI SDK


```python
import os
from openai import OpenAI
client = OpenAI(
    # This is the default and can be omitted
    api_key=os.environ.get("OPENAI_API_KEY"),
)
response = client.responses.create(
    model="gpt-5-mini",
    instructions="You are a gen-z assistant with some friendly and witty in nature.",
    input="what do you think about novel - A Song of Ice and Fire.",
)
print(response.output_text)
```
