Pydantic Wrapper around output



```python
from langchain import chat_models
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field
from langchain_core.prompts import ChatPromptTemplate
```

```python

openai_llm = chat_models.init_chat_model(model="gpt-5-mini", model_provider="openai", temperature=0.1)
# openai_llm = ChatOpenAI(model_name="gpt-5-mini", temperature=0.1)
```

```python
class llm_output_schema(BaseModel):
    movie_name: str = Field(..., description="The name of the movie")
    director: str = Field(..., description="Just The name of the director only")
    summary: str = Field(..., description="The short summary of the movie")
```

```python

structured_llm_ouput = openai_llm.with_structured_output(llm_output_schema)

res = structured_llm_ouput.invoke("Raja")

type(res)

res.director
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are gen-z assistant"),
    ("user", "Movie {movie_name} and Language {language}")
])

movie_name = input("Enter movie: ")
language = input("Enter language: ")
ready_prompt = prompt_template.invoke({"movie_name":movie_name, "language":language})

structured_llm_ouput = openai_llm.with_structured_output(llm_output_schema)
llm_output = structured_llm_ouput.invoke(ready_prompt)
type(llm_output.model_dump())
```
