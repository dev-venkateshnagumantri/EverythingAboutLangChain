Basic Input and Response Using LangChain

```python
from langchain_openai import ChatOpenAI
import os

openai_llm = ChatOpenAI(model="gpt-5-mini", temperature=0.2)
openai_llm.invoke("Bro, tell me a fun fact about - A Song of Ice and Fire.").content
```

```python
from langchain.chat_models import init_chat_model


openai_llm = init_chat_model(model="gpt-5-mini", temperature=0.2)
openai_llm.invoke("Bro, tell me a fun fact about show Game of Thrones.").content
```
