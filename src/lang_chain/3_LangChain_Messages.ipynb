LangChain Messages


```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_openai import ChatOpenAI

openai_model = ChatOpenAI(model="gpt-5-mini", temperature=0.2)
my_messages = [
    SystemMessage(content="You are gen-z assistant with some witty behaviour and always answers in fun way."),
    HumanMessage(content="Bro, tell me a fun fact about movie Bahubali, outside India.")
]

openai_model.invoke(my_messages).content
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain.chat_models import init_chat_model

openai_model = init_chat_model(model="gpt-5-mini", model_provider="openai", temperature=0.2)
my_messages = [
    SystemMessage(content="You are gen-z assistant"),
    HumanMessage(content="Bro, tell me a fun fact about mughal emperor aurangazeb")
]

openai_model.invoke(my_messages).content
```

```python
from langchain_core.prompts import PromptTemplate
from langchain.chat_models import init_chat_model

openai_model = init_chat_model(model="gpt-5-mini", model_provider="openai", temperature=0.2)


category = input("Enter category: ")
topic = input("Enter topic: ")

#prompt_template = PromptTemplate(template="Bro, give me a fun fact about {category} {topic}", input_variables=["category", "topic"])
prompt_template = PromptTemplate.from_template("Bro, give me a fun fact about {category} {topic}")

ready_prompt = prompt_template.invoke({"category":category, "topic":topic})

openai_model.invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

openai_model = init_chat_model(model="gpt-5-mini", model_provider="openai", temperature=0.4)

category = input("Enter category: ")
topic = input("Enter topic: ")

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are gen-z assistant"),
    ("user", "Hi Bro! Tell me a fun fact about {category} {topic}")
])

ready_prompt = prompt_template.invoke({"category":category, "topic":topic})

openai_model.invoke(ready_prompt).content

```
